# Importing

```python
from text2meta import Text2Meta
import pandas as pd

df = pd.read_csv('dataset/spam_or_ham.csv')
t2m = Text2Meta(data=df, text_column="sms", target="label")
```

# Using specifics metrics

```python
# Compute specific metrics
features = t2m.compute(metrics=["number_of_documents", "vocabulary_size"])
print(features)
```

# Using all metrics in specifics groups

```python
features = t2m.compute(groups=["general", "statistical"])
print(features)
```

```python
features = t2m.compute(groups=["general", "statistical", "info-theory", "model-based", "landmarking", "complexity"])
print(features)
```
